```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [100, 150, 130, 170, 200],
    'Profit': [20, 35, 30, 40, 50]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(6,4))
plt.plot(df['Month'], df['Sales'], marker='*', color='blue')

plt.title('Monthly Sales (Line Plot)')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(df['Month'], df['Sales'], color='green')

plt.title('Monthly Sales (Bar Plot)')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['Sales'], df['Profit'], color='red')

plt.title('Sales vs Profit (Scatter Plot)')
plt.xlabel('Sales')
plt.ylabel('Profit')

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['Sales'], bins=5, color='purple', edgecolor='black')

plt.title('Sales Distribution (Histogram)')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```
