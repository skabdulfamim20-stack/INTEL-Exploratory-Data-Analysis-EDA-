**Braod Casting in NumPy**

```python
#board casting in numpy
import numpy as np
a = np.array([1, 2, 3])
print(a + 5)
```

**Basic Operation in NumPy**

```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b)  #addition
print(a - b)  #subtraction
print(a * b)  #multiplication
print(a / b)  #division
```

**Shape Manipulation**

```python
import numpy as np
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print(b)
```

```python
import numpy as np
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print(a.reshape(3, 2))  #rehspe
print(a.flatten())  #Flattening
print(a.T)  #Transposing
```

**Array Manipulation in NumPy**

```python
import numpy as np
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)  #reshape
print(b)
print(b[0])   #indexing
```

**Structured Array**

```python
import numpy as np
data =  np.array([
                 (1, "shaik", 85.5),
                  (2, "Abdul", 90.0),
                 (3, "Famim", 78.0)
                 ], dtype = [('id', 'i4'), ('name', 'U10'), ('marks', 'f4')])
print(data)
```
