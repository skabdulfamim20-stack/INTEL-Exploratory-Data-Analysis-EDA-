```python
import pandas as pd
import matplotlib.pyplot as plt
data ={'product': ['A', 'B', 'C', 'D'], 'Sales': [50, 150, 200, 95] }
df = pd.DataFrame(data)
df.plot(kind='bar', x='product', y='Sales', color='red', title='sales by product')
plt.xlabel('product')
plt.ylabel('Sales')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {'Age': [20,22,25,26,28,45,68,19,75,60,62,65,70,75,80]}
df = pd.DataFrame(data)
df.plot(kind='hist', bins=8, color='lightgreen', title='Age Distribution')
plt.xlable('Age')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {'Year': [2005,2006,2007,2008,2009], 'Revenue': [100, 200, 210, 275, 300]}
df = pd.DataFrame(data)
df.plot(kind='line', x='Year', y='Revenue', color='blue', title='Revenue Over Years')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {'Height': [150,160,180,190], 'Weight': [50, 60, 65, 75]}
df = pd.DataFrame(data)
df.plot(kind='scatter', x='Height', y='Weight', color='red', title ='Height vs Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.show()
```
