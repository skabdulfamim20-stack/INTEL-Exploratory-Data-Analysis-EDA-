```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931]

plt.plot(yield_apples, marker= 'o')

plt.xlabel("Observation Number")
plt.ylabel("Apple Yield")
plt.title("Apple Yield Trend")

plt.show()
```

```python
years = [2010, 2011, 2012, 2013, 2014, 2015]
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931]

plt.plot(years, yield_apples, marker= 'o')

plt.xlabel("Year")
plt.ylabel("Apple Yield")
plt.title("Apple Yield Over Years")

plt.grid(True)
plt.show()
```

```python
years = [2010, 2011, 2012, 2013, 2014, 2015]
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931]
plt.plot(years, yield_apples)
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
plt.show()
```

```python
years = range(2000, 2010)
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375]
oranges = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898]

plt.plot(years, apples)
plt.plot(years, oranges)

plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

plt.show()
```

```python
plt.plot(years, apples, marker='o')
plt.plot(years, oranges, marker='x')

plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

plt.title("Crop Yields in kanto")
plt.legend(['Apples', 'Oranges']);
```

```python
years = range(2000, 2010)
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375]
oranges = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898]

plt.plot(years, apples, marker='s', c='b', ls='-', lw=2, ms=8, mew=2, mec='blue')
plt.plot(years, oranges, marker='o', c='r', ls='--', lw=3, ms=10, alpha=0.5)
```

```python
plt.plot(years, oranges, 'or')
plt.title("Yield of Oranges (tons per hectare)")
```

```python
yers = range(2000, 2012)
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375]

plt.scatter(years, apples)
plt.xlabel("Years")
plt.ylabel("Yield (tons per hectare)")
plt.title("Apple Yield Scatter Plot")
plt.show()
```

```python
years = range(2000, 2010)
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375]
oranges = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898]

plt.plot(years, apples, marker='o', color='green', label='Apples')
plt.plot(years, oranges, marker='s', color='orange', label='Oranges')

plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
plt.title('Apple and Orange Yields Comparison')
plt.legend()

plt.show()
```

```python
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375]
plt.hist(apples)
plt.xlabel("Yield")
plt.ylabel("Frequency")
plt.title("Distribution of Apple Yields")
```

```python
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375]
oranges = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898]

plt.hist(apples, alpha=0.7, label='Apples')
plt.hist(oranges, alpha=0.7, label='Oranges')

plt.xlabel('Yield (tons per hectare)')
plt.ylabel('Frequency')
plt.title('Histogram of Apple and Orange Yields')

plt.legend()

plt.show()
```
